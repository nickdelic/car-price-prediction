```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
```

```python
pot_podatkov = Path("../data/cars.csv")

podatki = pd.read_csv(pot_podatkov)

print("Podatki so uspešno naloženi.")
```

```python
display(podatki.head())
```

```python
display(
    podatki.sample(
        5,
        random_state=42
    )
)
```

```python
print("Število vrstic:", podatki.shape[0])
print("Število stolpcev:", podatki.shape[1])

print("\nStolpci:")
print(podatki.columns.tolist())
```

```python
podatki.info()
```

```python
print(podatki.dtypes)
```

```python
manjkajoce_vrednosti = (
    podatki
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print(manjkajoce_vrednosti)
```

```python
print(
    podatki["priceUSD"].describe()
)
```

```python
podatki["priceUSD"].plot(
    kind="hist",
    bins=50,
    figsize=(10, 5)
)

plt.title("Porazdelitev cen avtomobilov")
plt.xlabel("Cena v USD")
plt.ylabel("Število avtomobilov")
plt.show()
```

```python
display(
    podatki
    .sort_values("priceUSD")
    .head(20)
)
```

```python
display(
    podatki
    .sort_values(
        "priceUSD",
        ascending=False
    )
    .head(20)
)
```

```python
numericni_stolpci = [
    "priceUSD",
    "year",
    "mileage(kilometers)",
    "volume(cm3)"
]

display(
    podatki[
        numericni_stolpci
    ].describe()
)
```

```python
print(
    podatki["year"]
    .value_counts()
    .sort_index()
)
```

```python
podatki["mileage(kilometers)"].plot(
    kind="hist",
    bins=50,
    figsize=(10, 5)
)

plt.title("Porazdelitev kilometrine")
plt.xlabel("Kilometri")
plt.show()
```

```python
kategorijski_stolpci = [
    "make",
    "model",
    "condition",
    "fuel_type",
    "color",
    "transmission",
    "drive_unit",
    "segment"
]

for stolpec in kategorijski_stolpci:

    print("\n" + "=" * 60)
    print(stolpec)
    print("=" * 60)

    print(
        podatki[stolpec]
        .value_counts(
            dropna=False
        )
        .head(20)
    )
```
